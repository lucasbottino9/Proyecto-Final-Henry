```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../online_shoppers_intention.csv')

print("Cantidad de filas y columnas originales:")
print(df.shape)

df = df.drop_duplicates()
print("Cantidad de filas y columnas después de eliminar duplicados:")
print(df.shape)

print("Valores nulos por columna:")
print(df.isnull().sum())
```

# 2. Analisis de Datos (EDA) 

```python
df_procesado = pd.get_dummies(df)

df_procesado['Weekend'] = df_procesado['Weekend'].astype(int)
df_procesado['Revenue'] = df_procesado['Revenue'].astype(int)

print("Dataset listo:")
df_procesado.head(3)
```

# 3. Modelo y Evaluacion

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report
import matplotlib.pyplot as plt

x = df_procesado.drop(columns=['Revenue'])
y = df_procesado['Revenue']

X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

modelo_rf = RandomForestClassifier(random_state=42)
modelo_rf.fit(X_train, y_train)

y_pred = modelo_rf.predict(X_test)

exactitud = accuracy_score(y_test, y_pred)
print(f"Exactitud del modelo: {exactitud * 100:.2f}%")

print("\nReport de clasificación:")
print(classification_report(y_test, y_pred))
```
